## Converting khasra_number into int 

```python
col = "khasra_number"

print("Column:", col)
print("Data type:", df[col].dtype)
print("Unique values:", df[col].nunique())
print("Missing values:", df[col].isna().sum())
print("Sample values:")
print(df[col].value_counts(dropna=False).head(20))
```

```python
col = "khasra_number"

# Treat missing-value placeholders as NaN
df[col] = df[col].astype("string").str.strip()
df[col] = df[col].replace(
    r"(?i)^(na|n a|ns|nan|)$", pd.NA, regex=True
)

# Convert numeric values to numbers
df[col] = pd.to_numeric(df[col], errors="coerce")

# Check whether all remaining values are whole numbers
print("Non-integer values:",
      ((df[col].dropna() % 1) != 0).sum())

# Convert to nullable integer if all values are whole numbers
if ((df[col].dropna() % 1) == 0).all():
    df[col] = df[col].astype("Int64")

print("New data type:", df[col].dtype)
print(df[col].value_counts(dropna=False).head(20))
```

# removing unnecesary Colmuns

```python

target = "construction_cost"

# Columns with only one unique value (including missing values)
constant_cols = [
    col for col in df.columns
    if col != target and df[col].nunique(dropna=False) <= 1
]

# Potential identifier columns: review before dropping
id_like_cols = [
    col for col in df.columns
    if any(word in col.lower() for word in
           ["id", "khasra", "name", "number"])
    and col != target
]

print("Dataset shape:", df.shape)
print("\nConstant columns (usually safe to remove):")
print(constant_cols)

print("\nPotential ID/name/number columns (review first):")
print(id_like_cols)

print("\nTarget column:", target)
```
