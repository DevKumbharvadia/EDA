# Outliar detection

```python
df["scheme_depth"].describe()
```

```python
print("Smallest depth:", df["scheme_depth"].min())
print("Largest depth:", df["scheme_depth"].max())
print("Missing values:", df["scheme_depth"].isnull().sum())
```

```python
Q1 = df["scheme_depth"].quantile(0.25)
Q3 = df["scheme_depth"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = df[
    (df["scheme_depth"] < lower_bound) |
    (df["scheme_depth"] > upper_bound)
]

print("Number of potential outliers:", len(outliers))
```

# Removing misisng columns

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percent": df.isnull().mean() * 100
})

missing_summary = missing_summary.sort_values(
    "missing_percent", ascending=False
)

print(missing_summary[missing_summary["missing_count"] > 0].round(2))
```

```python
drop_cols = [
    "serial_no",
    "state_name",
    "scheme_in_command_area_reason",
    "name_of_mmi_project",
    "status_reason_permanently",
    "status_reason_temporarily",
    "not_in_use_years",
    "under_utilisation_reason",
    "dugwell_type"
]

print("Columns to remove:", len(drop_cols))
print(drop_cols)
```

```python
model_df = df.drop(columns=drop_cols).copy()

print("Original shape:", df.shape)
print("Model DataFrame shape:", model_df.shape)
```

# fixing data types

```python
model_df.dtypes
```

```python
model_df.head()
```

```python
object_cols = model_df.select_dtypes(include="object").columns

for col in object_cols:
    print(f"\n{col}")
    print(model_df[col].value_counts(dropna=False).head(5))
```

```python
import numpy as np

missing_labels = ["na", "NA", "Na", "n/a", "N/A", ""]

for col in model_df.select_dtypes(include="object").columns:
    model_df[col] = model_df[col].replace(
        missing_labels, np.nan
    )

print(model_df[["khasra_number", "location_particulars"]]
      .isna().sum())
```

```python
yes_no_cols = [
    "located_in_command_area",
    "only_recharge_ground_water",
    "under_utilised_selection",
    "drinking_water_purpose"
]

model_df[yes_no_cols] = model_df[yes_no_cols].replace({
    "Yes": 1,
    "No": 0
})

print(model_df[yes_no_cols].dtypes)
print(model_df[yes_no_cols].head())
```

```python
numeric_missing = (
    model_df.select_dtypes(include="number")
    .isna()
    .sum()
    .sort_values(ascending=False)
)

print(numeric_missing[numeric_missing > 0])
```

```python
pumping_cols = [
    "pump_operating_days_kharif_season",
    "pump_operating_days_rabi_season",
    "pump_operating_days_perennial_crop",
    "pump_operating_days_other_season",
    "avg_pumping_per_day_kharif_season",
    "avg_pumping_per_day_rabi_season",
    "avg_pumping_per_day_perennial_crop",
    "avg_pumping_per_day_other_season"
]

print("Rows missing any pumping value:",
      model_df[pumping_cols].isna().any(axis=1).sum())

print("Rows missing all pumping values:",
      model_df[pumping_cols].isna().all(axis=1).sum())
```

```python
import pandas as pd

# Columns that represent categories
categorical_cols = [
    "district_name",
    "block_tehsil_name",
    "village_name",
    "scheme_type",
    "tubewell_type",
    "ownership",
    "khasra_number",
    "location_particulars",
    "social_status",
    "gender",
    "commissioning_year",
    "finance_source1",
    "finance_source2",
    "scheme_status",
    "water_distribution_method",
    "lifting_device_type",
    "energy_source",
    "captured_in_previous_census"
]

# Convert category columns to pandas category dtype
for col in categorical_cols:
    model_df[col] = model_df[col].astype("category")

# Keep identifier-like column as text
model_df["s_no_of_scheme"] = model_df["s_no_of_scheme"].astype("string")

# Binary columns: preserve missing values
binary_cols = [
    "located_in_command_area",
    "only_recharge_ground_water",
    "under_utilised_selection",
    "drinking_water_purpose"
]

for col in binary_cols:
    model_df[col] = model_df[col].astype("Int64")

# Check the resulting types
print(model_df.dtypes.to_string())
```

```python
numeric_cols = model_df.select_dtypes(include="number").columns

negative_counts = (
    model_df[numeric_cols]
    .lt(0)
    .sum()
    .sort_values(ascending=False)
)

print(negative_counts[negative_counts > 0])
```

```python
unique_counts = model_df.nunique(dropna=False).sort_values()

print(unique_counts[unique_counts <= 1])
```

```python
cols_to_check = [
    "horse_power_of_lifting_device",
    "pump_operating_days_kharif_season",
    "pump_operating_days_rabi_season",
    "pump_operating_days_perennial_crop",
    "pump_operating_days_other_season"
]

for col in cols_to_check:
    values = model_df[col].dropna()
    decimals = values[values % 1 != 0]

    print(f"{col}:")
    print("  Non-missing values:", len(values))
    print("  Values with decimals:", len(decimals))
    print("  Examples:", decimals.head().tolist())
```
