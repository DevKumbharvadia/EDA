# Feature Engineering

```python
import numpy as np 
import pandas as pd

df = pd.read_csv("./data/05_Encoding.csv")
```

```python
df.shape
```

# Initial Dataset Inspection

```python
import pandas as pd
import numpy as np

print("=" * 60)
print("INITIAL FEATURE ENGINEERING INSPECTION")
print("=" * 60)

print("\n1. Dataset Shape:")
print(df.shape)

print("\n2. Duplicate Rows:")
print(df.duplicated().sum())

print("\n3. Missing Values:")
print(df.isnull().sum().sum())

print("\n4. Target Variable:")
print(df["construction_cost"].describe().round(2))

print("\n5. Numerical Feature Statistics:")
print(df.select_dtypes(include=np.number).describe().T.round(2).to_string())

print("\n6. Potential Index / Flag Columns:")
print([
    col for col in df.columns
    if col == "Unnamed: 0" or "Outlier" in col or "outlier" in col
])

print("\n7. Dataset Remains Unmodified:")
print(df.shape)
```

# Depth and Well Geometry Features

```python
import numpy as np
import pandas as pd

geometry_columns = ["scheme_depth", "scheme_diameter"]

print("=" * 60)
print("DEPTH AND WELL GEOMETRY ANALYSIS")
print("=" * 60)

print("\n1. Descriptive Statistics:")
print(df[geometry_columns].describe().round(3))

print("\n2. Correlation with Construction Cost:")
print(
    df[geometry_columns + ["construction_cost"]]
    .corr()["construction_cost"]
    .round(4)
)

print("\n3. Correlation Between Geometry Features:")
print(df[geometry_columns].corr().round(4))

print("\n4. Unique Values:")
for col in geometry_columns:
    print(f"{col}: {df[col].nunique()}")

print("\n5. Zero and Negative Values:")
print((df[geometry_columns] <= 0).sum())

print("\n6. Skewness:")
print(df[geometry_columns].skew().round(3))

print("\n7. Dataset Shape:", df.shape)
```

## Creating depth_diameter_ratio

```python
# Depth-to-Diameter Ratio

df["depth_diameter_ratio"] = (
    df["scheme_depth"] / df["scheme_diameter"]
)

print("New Feature: depth_diameter_ratio")

print("\nDescriptive Statistics:")
print(df["depth_diameter_ratio"].describe().round(4))

print("\nCorrelation with Construction Cost:")
print(
    df[["depth_diameter_ratio", "construction_cost"]]
    .corr()
    .iloc[0, 1]
    .round(4)
)

print("\nMissing Values:", df["depth_diameter_ratio"].isnull().sum())
print("Infinite Values:", np.isinf(df["depth_diameter_ratio"]).sum())
print("Updated Dataset Shape:", df.shape)
```

# Geographical Feature Engineering

```python
import pandas as pd

geo_columns = ["latitude", "longitude"]

print("=" * 60)
print("GEOGRAPHICAL FEATURE ANALYSIS")
print("=" * 60)

print("\n1. Descriptive Statistics:")
print(df[geo_columns].describe().round(4))

print("\n2. Correlation with Construction Cost:")
print(
    df[geo_columns + ["construction_cost"]]
    .corr()["construction_cost"]
    .round(4)
)

print("\n3. Unique Coordinate Values:")
for col in geo_columns:
    print(f"{col}: {df[col].nunique()}")

print("\n4. Geographic Coordinate Ranges:")
print("Latitude:", df["latitude"].min(), "to", df["latitude"].max())
print("Longitude:", df["longitude"].min(), "to", df["longitude"].max())

print("\n5. Missing Values:")
print(df[geo_columns].isnull().sum())

print("\n6. Dataset Shape:", df.shape)
```

## Crating geographic_cluster

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

geo_data = df[["latitude", "longitude"]]

scaler = StandardScaler()
geo_scaled = scaler.fit_transform(geo_data)

results = []

for k in range(2, 9):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(geo_scaled)

    score = silhouette_score(
        geo_scaled,
        labels,
        sample_size=10000,
        random_state=42
    )

    results.append((k, score))

print("SILHOUETTE SCORES")
print("-" * 35)

for k, score in results:
    print(f"K = {k}: {score:.4f}")

best_k = max(results, key=lambda x: x[1])[0]
print("\nBest K based on Silhouette Score:", best_k)
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

geo_data = df[["latitude", "longitude"]]

geo_scaler = StandardScaler()
geo_scaled = geo_scaler.fit_transform(geo_data)

geo_kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df["geographic_cluster"] = geo_kmeans.fit_predict(geo_scaled)

print("Geographic Clusters Created")

print("\nCluster Distribution:")
print(df["geographic_cluster"].value_counts().sort_index())

print("\nAverage Construction Cost by Cluster:")
print(
    df.groupby("geographic_cluster")["construction_cost"]
    .agg(["count", "mean", "median"])
    .round(2)
)

print("\nCorrelation with Target:")
print(
    df[["geographic_cluster", "construction_cost"]]
    .corr()
    .round(4)
)

print("\nUpdated Dataset Shape:", df.shape)
```

# Ownership Category Analysis

```python
ownership_columns = [
    col for col in df.columns
    if col.startswith("ownership_")
]

print("=" * 60)
print("OWNERSHIP CATEGORY ANALYSIS")
print("=" * 60)

print("\n1. Ownership Category Distribution:")
for col in ownership_columns:
    print(f"{col}: {df[col].sum()}")

print("\n2. Average Construction Cost by Ownership Category:")

for col in ownership_columns:
    group = df[df[col] == 1]["construction_cost"]

    print(
        f"{col}: Count={len(group)}, "
        f"Mean={group.mean():.2f}, "
        f"Median={group.median():.2f}"
    )

print("\n3. Dataset Shape:", df.shape)
```

# Cost-Related Feature Investigation

```python
cost_columns_to_drop = [
    "machinery_cost",
    "maintenance_cost",
    "construction_subsidy",
    "machinery_subsidy",
    "s_no_of_scheme"
]

df.drop(columns=cost_columns_to_drop, inplace=True)

print("Dropped Columns:")
print(cost_columns_to_drop)

print("\nUpdated Dataset Shape:", df.shape)

print("\nVerification:")
print("Remaining columns:", [
    col for col in cost_columns_to_drop if col in df.columns
])
```

# Feature Transformation

```python
import numpy as np
import pandas as pd

transform_candidates = [
    "scheme_depth",
    "culturable_command_area",
    "nearest_scheme_distance",
    "horse_power_of_lifting_device"
]

transform_candidates = [
    col for col in transform_candidates if col in df.columns
]

print("=" * 60)
print("FEATURE TRANSFORMATION ANALYSIS")
print("=" * 60)

print("\n1. Skewness:")
print(df[transform_candidates].skew().round(3))

print("\n2. Descriptive Statistics:")
print(df[transform_candidates].describe().round(3))

print("\n3. Zero and Negative Values:")
print((df[transform_candidates] <= 0).sum())

print("\n4. Correlation with Construction Cost:")
print(
    df[transform_candidates + ["construction_cost"]]
    .corr()["construction_cost"]
    .drop("construction_cost")
    .round(4)
)

print("\n5. Dataset Shape:", df.shape)
```

## Applying Log Transformation

```python
import numpy as np

df["culturable_command_area_log"] = np.log1p(
    df["culturable_command_area"]
)

print("Original Skewness:",
      round(df["culturable_command_area"].skew(), 4))

print("Transformed Skewness:",
      round(df["culturable_command_area_log"].skew(), 4))

print("\nCorrelation with Construction Cost:")
print(
    df[["culturable_command_area",
        "culturable_command_area_log",
        "construction_cost"]]
    .corr()["construction_cost"]
    .round(4)
)

print("\nUpdated Dataset Shape:", df.shape)
```
