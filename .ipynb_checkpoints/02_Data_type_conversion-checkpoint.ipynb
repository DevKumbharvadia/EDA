```python
import numpy as np
import pandas as pd

df = pd.read_csv("MI7_GROUNDWATERSCHEME_ASSAM.csv")
df.head()
```

```python
df.info()
```

## khasra_number is the wrong data type

```python
col = "khasra_number"

print("Column:", col)
print("Data type:", df[col].dtype)
print("Unique values:", df[col].nunique())
print("Missing values:", df[col].isna().sum())
print("Sample values:")
print(df[col].value_counts(dropna=False).head(20))
```

```python
df['khasra_number'] = (
    pd.to_numeric(df['khasra_number'], errors='coerce')
    .astype('Int64')
)

# converted khasra_number to int
df['khasra_number'].isna().sum()

print("New data type:", df[col].dtype)
print(df[col].value_counts(dropna=False).head(20))
```
